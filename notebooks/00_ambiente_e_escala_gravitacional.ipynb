# 00 — Ambiente e escala gravitacional

Este primeiro laboratório verifica o ambiente Python e explora uma escala fundamental da Relatividade Geral: o raio de Schwarzschild,

$$r_s = \frac{2GM}{c^2}. $$

O notebook utiliza apenas a base científica adotada no curso: **SymPy, NumPy, SciPy e Matplotlib**.

```python
import numpy as np
import sympy as sp
import scipy
import matplotlib.pyplot as plt
from scipy.constants import G, c

print(f"NumPy {np.__version__}")
print(f"SymPy {sp.__version__}")
print(f"SciPy {scipy.__version__}")
```

## 1. Dedução simbólica

No espaço-tempo de Schwarzschild, a componente temporal da métrica é

$$g_{tt}=-\left(1-\frac{2GM}{rc^2}\right).$$

O horizonte ocorre quando esse fator se anula.

```python
r, M, Gs, cs = sp.symbols("r M G c", positive=True)
f = 1 - 2 * Gs * M / (r * cs**2)
solucao = sp.solve(sp.Eq(f, 0), r)[0]
solucao
```

## 2. Valores para a Terra e o Sol

Vamos calcular $r_s$ em unidades SI e comparar essa escala com o raio físico dos corpos.

```python
def raio_schwarzschild(massa):
    """Retorna 2GM/c², em metros, para uma massa em quilogramas."""
    return 2 * G * massa / c**2

massa_terra = 5.9722e24       # kg
massa_sol = 1.98847e30        # kg

for nome, massa in {"Terra": massa_terra, "Sol": massa_sol}.items():
    rs = raio_schwarzschild(massa)
    print(f"{nome:5s}: r_s = {rs:.6g} m")
```

## 3. Como a escala depende da massa?

Usaremos massas medidas em unidades de massa solar. O gráfico log–log permite observar várias ordens de grandeza.

```python
massas_solares = np.logspace(-6, 10, 400)
raios_km = raio_schwarzschild(massas_solares * massa_sol) / 1e3

fig, ax = plt.subplots(figsize=(8, 4.8))
ax.loglog(massas_solares, raios_km, color="#1e5c83", linewidth=2.4)
ax.set(xlabel=r"Massa $M/M_\odot$", ylabel=r"Raio de Schwarzschild $r_s$ (km)")
ax.grid(True, which="both", alpha=0.22)
ax.set_title("A escala gravitacional cresce linearmente com a massa")
plt.show()
```
